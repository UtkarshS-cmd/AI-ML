## elif

````python
#elif
# marks =>
# marks < 30 --> "Low"
# marks between 30 and 70 --> "Avg"
# marks > 70 --> "High"

# using if-else only

marks = 20

if marks < 30:
  print("Low")
else:
  if (marks >= 30) and (marks <= 70):
    print("Avg")
  else:
    print("High")
````
