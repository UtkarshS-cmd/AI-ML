````python
for i in range(1,10):
  if i == 4:
    continue
  print(i)
````

````python
#infinite loop
i = 1
while i<= 10:

  if i == 4:
    continue

  print(i)
  i = i+1
````

````python
i = 1
while i<= 10:

  if i == 4:
    i = i+1
    continue

  print(i)
  i = i+1
````

````python
i = 1
while i<= 10:

  if i == 4:
    i = 5
    continue

  print(i)
  i = i+1
````
