Write a function to Print Grade of Every Student
Rules
90+
A
75+
B
60+


C
Else
D


```python
def calc_grade(marks):
  if marks > 90:
    grade = "A"
  elif marks > 75:
    grade = "B"
  elif marks > 60:
    grade = "C"
  else:
    grade = "D"
  return grade
```

```python
for student in students:
  print(student["marks"], calc_grade(student["marks"]))
```

```python
# Create Student Class (OOP) in the above format
class Student:
  def __init__(self, roll, name, age, course, marks):
    self.roll = roll
    self.name = name
    self.age = age
    self.course = course
    self.marks = marks
  def show_student(self):
    print(self.roll, self.name, self.age, self.course, self.marks)
```

```python
s1 = Student(106, "MJ", 30, "Dance", 100)
```

```python
s1.show_student()
```
