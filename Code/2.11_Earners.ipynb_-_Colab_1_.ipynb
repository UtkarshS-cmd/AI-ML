```python
#Write a function to return a list of the cube of every number in the given list.
def cube(lst):
  cb = []
  for n in lst:
    cb.append(n**3)
  return cb
```

```python
cube([5,7,3])
```

```python
def cb(n):
  return n**3
cube = []
lst = [4,6,2]
for elem in lst:
  cube.append(cb(elem))
print(cube)
```
