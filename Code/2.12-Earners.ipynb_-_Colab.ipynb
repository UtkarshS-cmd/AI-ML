```python
with open("student.csv", "w", newline="") as file:
  writer = csv.writer(file)
  writer.writerow(["Name", "Marks", "City"])
  writer.writerow(["Raj",  90, "Mumbai"])
  writer.writerow(["Rahul", 95, "Hyderabad"])
```
