## Coding Practice — Problem 1: Student Grade Analyzer

Write a Python program that takes marks of 5 subjects as input.

Your program should:
- Store the marks in a list.
- Calculate the average marks.
- Find the highest and lowest marks.
- Display the grade using:

| Average | Grade |
|---|---|
| 90+ | A |
| 75-89 | B |
| 60-74 | C |
| Below 60 | D |

```python
marks = []
for i in range(5):
    m = int(input())
    marks.append(m)
```

```python
s = sum(marks)
n = len(marks)
avg = s/n
print(avg)
```

```python
mn = min(marks)
mx = max(marks)
print(mn, mx)
```

```python
if avg >= 90:
    grade = "A"
elif avg >=75:
    grade = "B"
elif avg >= 60:
    grade = "C"
else:
    grade = "D"
print(grade)
```

### Final systematic code

```python
marks = []
for i in range(5):
    mark = int(input())
    marks.append(mark)

average = sum(marks) / len(marks)

print("Average :", average)
print("Highest :", max(marks))
print("Lowest :", min(marks))

if average >= 90:
    print("Grade : A")
elif average >= 75:
    print("Grade : B")
elif average >= 60:
    print("Grade : C")
else:
    print("Grade : D")
```

## Problem 2: Word Frequency Counter

Take a sentence as input.

Display how many times each word appears.

Example: `python is fun python is easy` → `python - 2, is - 2, easy - 1, fun - 1`

```python
s = input()
words = s.split(" ")
print(words)
```

```python
freq = {}
for word in words:
    if word not in freq:
        freq[word] = 1
    else:
        freq[word] += 1
print(freq)
```

## Problem 3: Remove Duplicate Names

Take names separated by spaces.

Remove duplicate names while preserving the first occurrence.

Example: `Rahul Amit Rahul Neha Amit` → `Rahul Amit Neha`

```python
s = input()
names = s.split()
print(names)
```

## Problem 4: Character Counter

Take a string as input from the user.

Display:
- Number of vowels
- Number of consonants
- Number of digits
- Number of spaces

Example input: `Python3 is Fun`

```python
s = input()
sl = s.lower()

vowels = 0
consonants = 0
digits = 0
spaces = 0

for c in sl:
    if c in "aeiou":
        vowels += 1
    elif c >= "a" and c <= "z":
        consonants += 1
    elif c >= "0" and c <= "9":
        digits += 1
    elif c == " ":
        spaces += 1

print("Vowels -", vowels)
print("Consonants -", consonants)
print("Digits -", digits)
print("Spaces -", spaces)
```

### Version 2

```python
text = input()
vowels = 0
consonants = 0
digits = 0
spaces = 0

for ch in text:
    if ch.lower() in "aeiou":
        vowels += 1
    elif ch.isalpha():
        consonants += 1
    elif ch.isdigit():
        digits += 1
    elif ch == " ":
        spaces += 1

print("Vowels :", vowels)
print("Consonants :", consonants)
print("Digits :", digits)
print("Spaces :", spaces)
```

## Problem 5: Shopping Bill

Take item names and prices for 5 products.

Store:
- item names in one list
- prices in another list

Display:
- total bill amount
- highest priced item
- lowest priced item

Example products: Apple 100, Milk 50, Bread 40, Rice 80, Juice 120.

```python
names = []
prices = []

for i in range(5):
    name = input()
    price = int(input())
    names.append(name)
    prices.append(price)

print(names)
print(prices)
```

```python
total_bill_amount = sum(prices)

mx = prices[0]
mx_prd = ""

mn = prices[0]
mn_prd = ""

for i in range(5):
    if prices[i] > mx:
        mx_prd = names[i]
        mx = prices[i]
    elif prices[i] < mn:
        mn_prd = names[i]
        mn = prices[i]

print(total_bill_amount)
print(mx, mx_prd)
print(mn, mn_prd)
```
