```python
class Counter:
    def __init__(self):
        self.count = 0
    def increase(self):
      print(self.count)
      self.count += 1
c = Counter()
for i in range(5):
    c.increase()
print(c.count)
```

```python
class Bank:
   def __init__(self, balance):
       self.balance = balance
   def withdraw(self, amount):
       if amount <= self.balance:
           self.balance -= amount
account = Bank(1000)
account.withdraw(200)
account.withdraw(300)
print(account.balance)
```
