Calculate the number of orders and percentage of total orders

represented by each order status.

A - 10 - 10/70

B - 20 - 20/70

C - 40 - 40/70

```python
select
Status,
count(*) as no_of_orders,
round(count(*)*100.0/(select count(*) from `e1.orders`),2) as
percentage_of_orders
from `e1.orders`
group by Status
order by no_of_orders desc
```

### PDF Page 2

Number of order-item records

Total sales value

```python
select
ProductID,
sum(Quantity) as total_quantity_sold,
count(*) as NoOfOrderItems,
sum(Total) as total_sale_Value
from `e1.order_items`
group by ProductID
order by total_quantity_sold desc
limit 5
```

Consider only order-item records where the total value of that

item is greater than Rs 2500.

For each product, calculate:

Number of such records

Total quantity

Total sales value

Display products having at least 3 qualifying records.

---
### PDF Page 3

```python
select
ProductID,
count(*) as NoOfOrderItems,
sum(Quantity) as total_quantity_sold,
sum(Total) as total_sale_Value
from `e1.order_items`
where total > 2500
group by ProductID
having NoOfOrderItems >= 3
```

For every warehouse, calculate:

Number of different products stored

Total stock

Average stock per product

Display only warehouses storing at least 5 different products.

Sort by total stock descending.

```python
select
WarehouseID,
count(distinct ProductID) as no_of_products,
sum(Stock) as total_stock,
round(avg(Stock),2) as avg_stock
from `e1.inventory`
group by WarehouseID
having  no_of_products >= 5
```
