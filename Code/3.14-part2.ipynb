```python
select
c.CategoryID,
c.CategoryName,
sum(o.Total) as total_sales
from `e1.products` as p
inner join `e1.order_items` as o
on p.ProductID = o.ProductID
inner join `e1.categories` as c
on c.CategoryID = p.CategoryID
group by c.CategoryID, c.CategoryName
order by total_sales desc;
```

```python
select
p.ProductID,
p.ProductName,
sum(oi.Total) as sales
from `e1.products` as p
left join `e1.order_items` as oi
on p.ProductID = oi.ProductID
group by p.ProductID, p.ProductName
```

```python
select
p.ProductID,
p.ProductName,
sum(oi.Total) as sales
from `e1.products` as p
left join `e1.order_items` as oi
on p.ProductID = oi.ProductID
where oi.ProductID is null
group by p.ProductID, p.ProductName
```

For every customer, show the number of orders they have placed,

including customers with zero orders.

```python
select
c.CustomerID,
c.Name,
count(o.OrderID) as no_of_orders
from `e1.customers` as c
left join `e1.orders` as o
on c.CustomerID = o.CustomerID
group by c.CustomerID, c.Name
order by no_of_orders asc
```

-- using right join

```python
select
c.CustomerID,
c.Name,
count(o.OrderID) as no_of_orders
from `e1.orders` as o
right join `e1.customers` as c
on c.CustomerID = o.CustomerID
```
