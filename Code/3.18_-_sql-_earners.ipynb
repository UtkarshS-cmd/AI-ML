### PDF Page 3

Number of different customers served --> orders

Display only employees who have handled more than 5 orders. -->

having

Sort employees by number of orders handled from highest to

lowest. --> order by

```python
select
e.EmployeeID,
e.Name,
e.Department,
count(distinct o.OrderID) as total_orders,
count(distinct o.CustomerID) as different_customers
from `e1.orders` as o
inner join `e1.employees` as e
on o.EmployeeID = e.EmployeeID
group by
e.EmployeeID,
e.Name,
e.Department
having total_orders > 5
order by total_orders desc
```

### PDF Page 4

For every city, calculate:

Number of customers --> customers

Number of orders --> orders

Total sales value --> order_items

Average order-item value --> order_items

Display only cities where the total sales value is greater than

Rs 3,00,000. --> having

Sort cities by total sales in descending order. --> order by

```python
select
c.City,
count(distinct c.CustomerID) as no_of_customers,
count(distinct o.OrderID) as no_of_orders,
sum(oi.Total) as total_sales,
round(avg(oi.Total),2) as avg_item_value
from `e1.customers` as c
inner join `e1.orders` as o
on c.CustomerID = o.CustomerID
inner join `e1.order_items` as oi
on o.OrderID = oi.OrderID
group by c.City
having total_sales> 300000
order by total_sales desc
```

### PDF Page 5

For every product that has been sold, calculate:

ProductID --> product

ProductName --> product

Number of orders containing the product --> order_items

Total quantity sold --> order_items

Average selling price --> order_items

Display only products whose average selling price is greater

than Rs 15,000. --> having

Show the highest average selling price first. --> order by

```python
select
p.ProductID,
p.ProductName,
count(distinct oi.OrderID) as no_of_orders,
sum(oi.Quantity) as total_qty,
avg(oi.SellingPrice) as avg_selling_price
from `e1.products` as p
inner join `e1.order_items` as oi
on p.ProductID = oi.ProductID
group by
p.ProductID,
p.ProductName
having avg_selling_price > 15000
```

For every category, calculate:

CategoryName --> category

Number of products --> product

Minimum MRP --> product

Maximum MRP --> product

Difference between maximum and minimum MRP --> product

Display only categories where the difference between the maximum

and minimum MRP is greater than Rs 10,000.

```python
select
c.CategoryName,
count(distinct p.ProductID) as no_of_products,
min(p.MRP) as min_mrp,
max(p.MRP) as max_mrp,
max(p.MRP) - min(p.MRP) as mrp_diff
from `e1.categories` as c
inner join `e1.products` as p
on c.CategoryID = p.CategoryID
group by  c.CategoryName
having mrp_diff > 100
```
