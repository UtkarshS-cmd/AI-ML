How many different products has each customer purchased?

```python
select
c.CustomerID,
c.Name,
count(distinct oi.ProductID) as diff_prds
from `e1.customers` as c
inner join `e1.orders` as o
on c.CustomerID = o.CustomerID
inner join `e1.order_items` as oi
on o.OrderID = oi.OrderID
group by
c.CustomerID,
c.Name
order by diff_prds desc
```

### PDF Page 8

Display:

CategoryID

CategoryName

Total Sales

Number of different products sold

Only include categories where the total sales are greater than

Rs 1,00,000.

Sort the result from highest to lowest total sales.

```python
select
ct.CategoryID,
ct.CategoryName,
sum(oi.Total) as total_sales,
count(distinct p.ProductID) as no_of_prod
from `e1.categories` as ct
join `e1.products` as p
on ct.CategoryID=p.CategoryID
join `e1.order_items` as oi
on oi.ProductID=p.ProductID
group by
ct.CategoryID,
ct.CategoryName
having total_sales>100000
order by total_sales desc
```
