How many products are there in each category?


```python
select
CategoryID,
count(*) as total_count
from `e1.products`
group by CategoryID
```

How many units of each product have been sold?


```python
select *
from `e1.order_items`
limit 5
```

```python
select
ProductID,
sum(Quantity) as total_qty
from `e1.order_items`
group by ProductID
```

```python
select *
from `e1.order_items`
```

What is the total sales value generated by each product?


```python
select
ProductID,
sum(total) as total_sales
from `e1.order_items`
group by ProductID
```

What was the average selling price at which each product was sold?


```python
select
ProductID,
avg(SellingPrice) as avg_selling_price
from `e1.order_items`
group by ProductID
```

What was the lowest and highest selling price recorded for each product?


```python
select *
from `e1.order_items`
limit 5
```

```python
select
ProductID,
min(SellingPrice) as lowest_sp,
max(SellingPrice) as highest_sp
from `e1.order_items`
group by ProductID
```

for each of the order status, what is the number of orders for each of the customer?


```python
select *
from `e1.orders`
limit 5
```

```python
select
CustomerID,
status,
count(*) as total_count
from `e1.orders`
group by CustomerID, Status
```

```python
select
CustomerID,
status,
count(*) as total_count
from `e1.orders`
group by Status, CustomerID
```

```python
How many inventory records are maintained for each warehouse?
select *
from `e1.inventory`
limit 5
```

```python
select
WarehouseID,
count(*) as no_of_records
from `e1.inventory`
group by WarehouseID
```

How many Orders are handled by each of the employees for each Status?


```python
select
EmployeeID,
Status,
count(*) as total_orders
from `e1.orders`
group by EmployeeID, Status
```

How many products in each category have an MRP above ₹20,000?


```python
select *
from `e1.products`
```

```python
select
CategoryID,
count(*) as total_products
from `e1.products`
where MRP > 2000
group by CategoryID
```

What is the total quantity sold for each product when quantity is greater than 1?


```python
select *
from `e1.order_items`
limit 5
```

```python
select
ProductID,
sum(Quantity) as total_qty
from `e1.order_items`
where Quantity > 1
group by ProductID
```

What is the average payment for each payment method among successful payments?


```python
select
Method,
round(avg(amount), 2) as avg_amt
from `e1.payments`
where status = "Success"
group by Method
```
