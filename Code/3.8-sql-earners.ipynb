Display the payment methods with the highest total payment
amount at the top?


```python
select *
from `e1.payments`
limit 5
```

```python
select
Method,
sum(Amount) as total_payment
from `e1.payments`
group by Method
order by total_payment desc
```

Display the supplierIDs with
Highest number of products first.


If tied, highest average selling price first.


```python
select *
from `e1.products`
limit 5
```

```python
select
SupplierID,
count(*) as no_of_products,
avg(SellingPrice) as avg_sp
from `e1.products`
group by SupplierID
order by no_of_products desc, avg_sp desc
```

Sort products by category first, and within each category sort
by selling price from highest to lowest.


```python
select
ProductID,
CategoryID,
SellingPrice
from `e1.products`
order by CategoryID, SellingPrice desc
```

1. Take products
↓
2. Keep MRP > 1000
↓
3. Create supplier groups
↓
4. Count products
↓
5. Keep suppliers with > 2
↓
6. Sort highest count first
↓
7. Show top 5


```python
select
SupplierID,
count(*) as prd_cnt
from `e1.products`
where MRP > 1000
group by SupplierID
having prd_cnt > 2
order by prd_cnt desc
limit 5
```

Give me individual products whose MRP is greater than ₹20,000.


```python
select *
from `e1.products`
where MRP > 20000
```

Which suppliers have more than 3 products?


```python
select
SupplierID,
count(*) as product_cnt
from `e1.products`
group by SupplierID
having product_cnt > 3
```

Which customers have placed more than 5 orders?

```python
select
count(*) as no_of_orders
from `e1.orders`
group by CustomerID
having no_of_orders > 5
```
