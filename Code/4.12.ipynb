## Bank Account Transactions

A bank account has the following transactions:

```python
transactions = np.array([
 5000, -2000, 15000, -4500, 7000,
-12000, 25000, -3000, 8000, -18000,
4500, 22000, -6000, 9000, -2500
])
```

Positive values represent deposits and negative values represent withdrawals.

Tasks:
- Extract deposits.
- Extract withdrawals.
- Calculate total deposits and withdrawals.
- Calculate net transaction amount.
- Calculate absolute transaction values.
- Find transactions whose absolute value is above the average absolute transaction.
- Find the largest deposit.
- Find the largest withdrawal.
- Create labels `"Credit"` and `"Debit"`.
- Find percentage of transactions whose absolute value is greater than ₹10,000.
- Reshape the data into 3 × 5.
- Calculate total transaction value for each row.

````python
transactions = np.array([
 5000, -2000, 15000, -4500, 7000,
-12000, 25000, -3000, 8000, -18000,
4500, 22000, -6000, 9000, -2500
])
````

````python
deposits = transactions[transactions > 0]
deposits
````

````python
withdrawals = transactions[transactions < 0]
withdrawals
````

````python
total_d = np.sum(deposits)
total_w = np.sum(withdrawals)
total_d
total_w
````

````python
net_amt = np.sum(transactions)
net_amt
````

````python
abs_t = np.abs(transactions)
abs_t
````

````python
# Find transactions whose absolute value is above the average absolute transaction.
avg_abs_t = np.mean(abs_t)
avg_abs_t

# abs_t[abs_t > avg_abs_t]
transactions[abs_t > avg_abs_t]
````

````python
# Find the largest deposit.
np.max(transactions)
````

````python
# Find the largest withdrawal.
np.min(transactions)
````

````python
# Create labels "Credit" and "Debit".
labels = np.where(transactions > 0, "Credit", "Debit")
labels
````

````python
# Find percentage of transactions whose absolute value is greater than ₹10,000.
abs10k = abs_t[abs_t > 10000]
m = len(abs10k)
n = len(abs_t)
perc = m*100/n
perc
````

````python
# Reshape the data into 3 × 5.
t_new = transactions.reshape(3,5)
t_new
````

````python
# Calculate total transaction value for each row.
row_sm = np.sum(t_new, axis = 1)
row_sm
````

## Factory Production

A factory has four machines operating for six days:

```python
production = np.array([
    [120, 135, 128, 145, 150, 138],
    [110, 125, 118, 130, 142, 135],
    [160, 155, 170, 165, 180, 175],
    [95, 105, 100, 115, 108, 120]
])
```

Tasks:
- Calculate total production by machine.
- Calculate total production by day.
- Calculate average production of each machine.
- Find the machine with the highest average production.
- Find the day with the lowest production.
- Find values above the overall average.
- Increase production values below 120 by 5%.
- Extract alternate days.
- Extract the last 3 days for machines 1 and 3.
- Create `"High"` and `"Normal"` labels based on the overall average.

````python
production = np.array([
    [120, 135, 128, 145, 150, 138],
    [110, 125, 118, 130, 142, 135],
    [160, 155, 170, 165, 180, 175],
    [95, 105, 100, 115, 108, 120]
])
````

````python
machine_total_p = np.sum(production, axis = 1)
machine_total_p
````

````python
# Calculate total production by day.
day_total_p = np.sum(production, axis = 0)
day_total_p
````

````python
# Calculate average production of each machine.
machine_avg = np.mean(production, axis = 1)
machine_avg
````

````python
# Find the machine with the highest average production.
ind = np.argmax(machine_avg)
ind
production[ind]
````

````python
# Find the day with the lowest production.
day_total_p
np.argmin(day_total_p)
````

````python
# Find values above the overall average.
overall_avg = np.mean(production)
overall_avg
production[production > overall_avg]
````

````python
# Increase production values below 120 by 5%.
production = np.where(production < 120, production*1.05, production)
production
````

````python
# Extract alternate days.
production[:,::2]
````

````python
# Extract the last 3 days for machines 1 and 3.
production[[0,2]]
production[[0,2], -3:]
````

````python
# Create "High" and "Normal" labels based on the overall average.
status = np.where(
    production > overall_avg,
    "High",
    "Normal"
)
print(status)
````

## Loan Portfolio

```python
loan_amount = np.array([
    100000, 250000, 500000, 750000,
    120000, 350000, 900000, 450000
])

interest_rate = np.array([
    8.5, 9.2, 8.8, 10.1,
    7.9, 9.5, 10.5, 8.7
])
```

Tasks:
- Calculate annual interest for every loan.
- Calculate total interest.
- Calculate average interest.
- Find the loan generating the highest interest.
- Find loans above ₹400,000 with interest rate above 9%.

### HW
- Categorize loans: Rate >= 10 → High; Rate >= 9 → Medium; Otherwise → Low.
- Round interest to 2 decimal places.
- Find median loan amount.
- Find loans above median amount AND interest above average interest.
- Calculate standard deviation of interest rates.

````python
loan_amount = np.array([
    100000, 250000, 500000, 750000,
    120000, 350000, 900000, 450000
])

interest_rate = np.array([
    8.5, 9.2, 8.8, 10.1,
    7.9, 9.5, 10.5, 8.7
])
````

````python
# Calculate annual interest for every loan.
annual_interest = (loan_amount*interest_rate)/100
annual_interest
````

````python
# Calculate total interest.
total_int = np.sum(annual_interest)
total_int
````

````python
# Calculate average interest.
avg_int = np.mean(annual_interest)
avg_int
````

````python
# Find the loan generating the highest interest.
np.argmax(annual_interest)
````

````python
# Find loans above ₹400,000 with interest rate above 9%.
criteria = (loan_amount > 400000) & (interest_rate > 9)
print(criteria)
loan_amount[criteria]
````
