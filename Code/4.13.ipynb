## Delivery Locations

The coordinates of eight delivery locations are given as `[x, y]`.

Tasks:
- Calculate distance from the origin: `√(x² + y²)`.
- Extract x and y coordinates.
- Calculate distance for every location.
- Round distances to 2 decimals.
- Find nearest and farthest location.
- Find average distance.
- Find locations above average distance.
- Categorize: >= 30 → Long, >= 15 → Medium, otherwise → Short.
- Find percentage of Long-distance deliveries.
- Flatten the coordinate array.
- Find the size of the flattened array.

```python
points = np.array([
    [3, 4],
    [5, 12],
    [8, 15],
    [7, 24],
    [9, 40],
    [12, 35],
    [15, 20],
    [6, 8]
])
```

```python
# Extract x and y coordinates.
x = points[:, 0]
y = points[:, 1]
x, y
```

```python
# Calculate distance for every location.
dist = np.sqrt(np.square(x) + np.square(y))
dist
```

```python
# Find nearest and farthest location.
np.min(dist)
```

```python
np.max(dist)
```

```python
# Find average distance.
avg_dist = np.mean(dist)
avg_dist
```

```python
# Find locations above average distance.
points[dist > avg_dist]
```

```python
# Categorize:
# if >= 30 → Long
# elseif >= 15 → Medium
# Otherwise → Short

category = np.where(
    dist >= 30,
    "Long",
    np.where(dist >= 15, "Medium", "Short")
)
category
```

```python
# Find percentage of Long-distance deliveries.
long_dist = category[category == "Long"]
long_dist

m = len(long_dist)
n = len(dist)

perc = (m * 100) / n
perc
```

```python
points
```

```python
# Flatten the coordinate array.
points.flatten()
```

```python
# Find the size of the flattened array.
len(points.flatten())
```

### `axis` in concatenation

The source notes describe:
- `axis = 1` as concatenating horizontally (adding columns).
- `axis = 0` as concatenating vertically (adding rows).

```python
a = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

b = np.array([
    [7, 8, 9],
    [10, 11, 12]
])
```

```python
res = np.concatenate((a, b), axis=0)
res
```

```python
res = np.concatenate((a, b), axis=1)
res
```
