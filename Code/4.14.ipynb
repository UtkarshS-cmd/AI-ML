## HW — Inventory

Inventory across four stores and five products.

```python
import numpy as np

inventory = np.array([
    [12, 45, 8, 32, 20],
    [5, 60, 15, 25, 10],
    [30, 18, 6, 40, 22],
    [9, 55, 12, 28, 7]
])

inventory
```

Tasks:
- Calculate total inventory per store.
- Calculate total inventory per product.
- Find the store with highest inventory.
- Find the product with lowest total inventory.
- Extract inventory values below 10.
- Count items that require replenishment.
- Create `"Reorder"` / `"OK"` labels.
- Increase inventory below 10 by 20%.
- Round the resulting inventory.
- Find median inventory.
- Extract the last two products.
- Flatten the inventory and calculate standard deviation.

```python
# Calculate total inventory per store.
total_inv = np.sum(inventory, axis=1)
total_inv
```

```python
# Calculate total inventory per product.
prd_inv = np.sum(inventory, axis=0)
prd_inv
```

```python
# Find the store with highest inventory.
high_store = np.argmax(total_inv)
ans = high_store + 1
ans
```

```python
# Find the product with lowest total inventory.
low_inv_prd = np.argmin(prd_inv)
ans = low_inv_prd + 1
ans
```

```python
# Extract inventory values below 10.
low_inv = inventory[inventory < 10]
low_inv
```

```python
# Count items that require replenishment.
n = len(low_inv)
n
```

```python
m = np.sum(inventory < 10)
m
```

```python
# Create "Reorder" / "OK" labels.
label = np.where(inventory < 10, "Reorder", "OK")
label
```

```python
# Increase inventory below 10 by 20%.
new_inv = np.where(inventory < 10, inventory * 1.2, inventory)
new_inv
```

```python
# Round the resulting inventory.
new_inv = np.round(new_inv)
new_inv
```

```python
# Find median inventory.
np.median(new_inv)
```

```python
flat_inv = new_inv.flatten()
flat_inv
```

```python
np.sort(flat_inv)
```

```python
# Extract the last two products.
new_inv[:, -2:]
```

```python
# Flatten the inventory and calculate standard deviation.
flat_inv
np.std(flat_inv)
np.std(new_inv)
```

### Concatenation error example

The source intentionally demonstrates that `axis=1` requires the arrays to have matching sizes along dimension 0.

```python
a = np.array([
    [1, 2],
    [3, 4],
    [6, 7]
])

b = np.array([
    [5, 6],
    [7, 8]
])

# This raises ValueError because the row counts differ.
a1 = np.concatenate((a, b), axis=1)
```

```python
b1 = np.concatenate((a, b), axis=0)
b1
```

### `vstack` and `hstack`

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

result = np.vstack((a, b))
print(result)
```

```python
# This raises ValueError because the vectors have different lengths.
a = np.array([1, 2, 3, 7])
b = np.array([4, 5, 6])

result = np.vstack((a, b))
print(result)
```

```python
a = np.array([1, 2, 3, 7])
b = np.array([4, 5, 6])

result = np.hstack((a, b))
print(result)
```

## Reshaping and Stacking Examples

The source notes include examples for quarterly sales, weekly inventory, and combining employee data.

```python
# jan-mar:
# apr - june:

#     W1   w2   w3   w4
# i1
# i2
# i3
# i4
# i5
# i6

# sales of 3 products
# jan
# feb
# march
# qtr1

jan_sales = np.array([70, 80, 90])
feb_sales = np.array([65, 75, 85])
mar_sales = np.array([88, 92, 95])

overall_sales = np.vstack((jan_sales, feb_sales, mar_sales))
print(overall_sales)
```

```python
qty_w1 = np.array([70, 80, 90, 100, 20, 90, 70]).reshape(-1, 1)
qty_w2 = np.array([70, 80, 90, 100, 20, 90, 70]).reshape(-1, 1)
qty_w3 = np.array([70, 80, 90, 100, 20, 90, 70]).reshape(-1, 1)
qty_w4 = np.array([70, 80, 90, 100, 20, 90, 70]).reshape(-1, 1)

overall_inventory = np.hstack((qty_w1, qty_w2, qty_w3, qty_w4))
print(overall_inventory)
overall_inventory.shape
```

```python
ids = np.array([101, 102, 103]).reshape(-1, 1)
salary = np.array([50000, 65000, 72000]).reshape(-1, 1)
experience = np.array([3, 5, 7]).reshape(-1, 1)

df = np.hstack((ids, salary, experience))
df
```

```python
ids = np.array([101, 102, 103])
salary = np.array([50000, 65000, 72000])
experience = np.array([3, 5, 7])

df1 = np.column_stack((ids, salary, experience))
df1
```

```python
quantity = np.array([2, 5, 3, 4])
price = np.array([100, 200, 150, 300])

# revenue
revenue = quantity * price
revenue
```

```python
df = np.column_stack((quantity, price, revenue))
df
```
