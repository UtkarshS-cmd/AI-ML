```python
import numpy as np

prices = np.array([
    820, 835, 810, 850, 875,
    860, 890, 910, 895, 930,
    945, 920, 960, 975, 990
])
```

### Price analysis tasks

- Calculate percentage change between consecutive days.
- Round percentage changes to 2 decimals.
- Find days where price increased by more than 2%.
- Find maximum, minimum and range.
- Reshape prices into 3 × 5.
- Find average price of each row.

```python
diff = prices[1:] - prices[:-1]
diff
```

```python
# Calculate percentage change between consecutive days.
perc_change = np.round((diff * 100) / prices[:-1], 2)
perc_change
```

```python
# Find days where price increased by more than 2%.
high_change_days = perc_change[perc_change > 2]
high_change_days
```

```python
# Find days where price increased by more than 2%.
high_change_days = np.where(perc_change > 2, "Yes", "No")
high_change_days
```

```python
for i in range(len(diff)):
    if high_change_days[i] == "Yes":
        print(i)
```

```python
# Find maximum, minimum and range.
mx = np.max(perc_change)
mn = np.min(perc_change)
range_value = mx - mn

print("Range:", range_value)
print("Max:", mx)
print("Min:", mn)
```

```python
# Reshape prices into 3 × 5.
new_price = prices.reshape(3, 5)
new_price
```

```python
# Find average price of each row.
avg_row_price = np.mean(new_price, axis=1)
avg_row_price
```

## Campaign Leads Analysis

```python
# A company records leads generated by four campaigns across five channels:
leads = np.array([
    [12000, 15000, 11000, 18000, 21000],
    [14000, 13000, 17000, 19000, 22000],
    [10000, 16000, 15000, 20000, 23000],
    [18000, 17000, 19000, 21000, 25000]
])

leads.shape
```

## Stacking Arrays

```python
jan = np.array([
    [101, 5000],
    [102, 7000],
    [103, 4500]
])

feb = np.array([
    [104, 6200],
    [105, 8100],
    [106, 5300]
])
```

```python
# Vertical stacking
sales_data = np.vstack((jan, feb))
print(sales_data)
```

```python
# Horizontal stacking
sales_data = np.hstack((jan, feb))
print(sales_data)
```

```python
sales_data = np.array([
    [101, 5000],
    [102, 7000],
    [103, 4500]
])

bonus = np.array([
    [500],
    [700],
    [450]
])

np.hstack((sales_data, bonus))
```

### `vstack` shape error

The PDF demonstrates that `np.vstack((sales_data, bonus))` raises a `ValueError` because the arrays have different numbers of columns: `sales_data` has 2 columns while `bonus` has 1 column.
