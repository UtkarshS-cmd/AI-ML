```python
quantity = orders[:, 1]
price = orders[:, 2]
rating = orders[:, 3]
```

```python
# Find orders where quantity >= 5 AND rating >= 3.
criteria = (quantity >= 5) & (rating >= 3)
criteria
```

```python
orders[criteria]
```

```python
med_price = np.median(price)
med_price
```

```python
# Find orders where price is above median OR rating is below 4.
criteria = (price > med_price) | (rating < 4)
criteria
```

```python
orders[criteria]
```

```python
# Give 10% discount to products whose price is above ₹1,000.
new_price = np.where(price > 1000, price * 0.9, price)
new_price
```

```python
# Round discounted prices to 2 decimals.
new_price = np.round(new_price, 2)
new_price
```

```python
# Calculate standard deviation of price.
std_price = np.std(price)
std_price
```

```python
# Find 25th and 75th percentile of price.
p25 = np.percentile(price, 25)
p25
```

```python
p75 = np.percentile(price, 75)
p75
```

```python
# Find prices above the 75th percentile.
price75 = price[price > p75]
price75
```

```python
# Find percentage of orders with rating >= 3.
ord3 = rating[rating >= 3]
perc = len(ord3) * 100 / len(rating)
perc
```
