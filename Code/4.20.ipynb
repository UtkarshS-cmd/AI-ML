## 1. Cartesian product of two lists — using loops

```python
numbers1 = [1, 2, 3]
numbers2 = [10, 20]

pair = []
for n1 in numbers1:
    for n2 in numbers2:
        pair.append((n1, n2))
pair
```

## 2. Cartesian product — list comprehension

```python
pair = [(n1, n2) for n1 in numbers1 for n2 in numbers2]
pair
```

## 3. All combinations of T-shirt colors and sizes

```python
colors = ["Red", "Blue", "Black"]
sizes = ["S", "M", "L"]
products = [[c, s] for c in colors for s in sizes]
products
```

## 4. Generate pairs whose sum is greater than 12 — using loops

```python
numbers1 = [1, 2, 3]
numbers2 = [10, 20]
ans = []
for n1 in numbers1:
    for n2 in numbers2:
        if n1 + n2 > 12:
            ans.append((n1, n2))
ans
```

## 5. Generate pairs whose sum is greater than 12 — list comprehension

```python
ans = [(n1, n2) for n1 in numbers1 for n2 in numbers2 if n1 + n2 > 12]
ans
```

## 6. Pairs from 1 to 5 with different values and an even product

```python
pairs = [
    (n1, n2)
    for n1 in range(1, 6)
    for n2 in range(1, 6)
    if n1 != n2 and (n1 * n2) % 2 == 0
]
pairs
```

## 7. Print rows from a nested list

```python
marks = [
    [75, 80, 90],
    [60, 85, 70],
    [95, 88, 92]
]
for r in marks:
    print(r)
```

## 8. Print each element in a row

## 9. Flatten a nested list — using loops

```python
marks = [
    [75, 80, 90],
    [60, 85, 70],
    [95, 88, 92]
]
ans = []
for r in marks:
    for e in r:
        ans.append(e)
ans
```

## 10. Flatten a nested list — list comprehension

```python
flat = [e for r in marks for e in r]
flat
```

## 11. Extract passing marks (>= 75) from every row

```python
flat = [e for r in marks for e in r if e >= 75]
flat
```

## 12. Increase every number by 5

```python
marks = [
    [75, 80, 90],
    [60, 85, 70],
    [95, 88, 92]
]
flat = [e + 5 for r in marks for e in r]
flat
```

## 13. Replace negative values with zero in a matrix

```python
data = [
    [10, -5, 20],
    [-3, 15, -8],
    [25, 0, 12]
]
ans = [e if e >= 0 else 0 for r in data for e in r]
ans
```

## 14. Generate all possible three-digit combinations

```python
digits = [1, 2, 3]
ans = [(a, b, c) for a in digits for b in digits for c in digits]
print(ans)
len(ans)
```

## 15. Generate combinations where all three values differ

```python
digits = [1, 2, 3]
ans = [
    (a, b, c)
    for a in digits
    for b in digits
    for c in digits
    if a != b and b != c and a != c
]
ans
```

## 16. Print three-digit combinations with nested loops

```python
digits = [1, 2, 3]
for a in digits:
    for b in digits:
        for c in digits:
            print(a, b, c)
    print("*" * 10)
```

## 19. Apply tax to product prices
Apply 18% tax to products priced above ₹1,000 and 5% tax to all other products.

```python
prices = [500, 1200, 2500, 800, 1500]
new_prices = [p * 1.18 if p > 1000 else p * 1.05 for p in prices]
new_prices
```

## 22. Convert full names into lowercase usernames
Join first and last names with a dot.

Input names: `Aman Sharma`, `Riya Verma`, `Karan Patel`.

## 23. Keep each value only once, preserving first appearance

```python
numbers = [4, 2, 4, 7, 2, 9, 7, 10]

# TODO: Create a list of unique values while preserving order.
unique_numbers = []
for n in numbers:
    if n not in unique_numbers:
        unique_numbers.append(n)
unique_numbers
```
