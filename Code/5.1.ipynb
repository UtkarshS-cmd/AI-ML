```python
# Convert full names into lowercase usernames by joining the first and last names with a dot.
names = [
 "Aman Sharma",
 "Riya Verma",
 "Karan Patel"
]
```

```python
ans = []
for name in names:
 name = name.lower()
 nm = name.split()
 new_nm = nm[0]+"."+nm[1]
 ans.append(new_nm)
print(ans)
```

```python
# Create a new list containing each value only once, preserving the order of its first appearance.
numbers = [4, 2, 4, 7, 2, 9, 7, 10]
ans = []
for n in numbers:
 if n not in ans:
   ans.append(n)
print(ans)
```

```python
# Given a list of integers, determine whether the list is sorted in ascending order.
# Do not use sort() or sorted().
# Input: [10, 20, 30, 40]
# Output: Sorted
# Input: [10, 30, 20, 40]
# Output: Not Sorted
# a = [10, 20, 30, 40]
a = [10, 30, 20, 40]
n = len(a)
flag = 0
for i in range(n-1):
 if a[i] > a[i+1]:
 flag = 1
 break
if flag == 1:
 print("No")
else:
 print("yes")
```
