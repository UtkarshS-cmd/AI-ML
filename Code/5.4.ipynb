Find the minimum Absolute Sum of Elements

PRoblm Statement: Find 2 elemetns at diff indices whose sum has the smallest absolute value

Example: [-8,4,5,-2,10]
output: 2 --> (-8,10) or (4, -2)

```python

arr = [-8, 4, 5, -2, 10]

def find_min_sum(arr):
    min_sum = arr[0] + arr[1]
    n = len(arr)

    for i in range(n):
        for j in range(i + 1, n):
            current_sum = arr[i] + arr[j]

            if abs(current_sum) < abs(min_sum):
                min_sum = current_sum

    return min_sum

print(find_min_sum(arr))
```

Find the Number of Pairs with a difference of K

Problem statement:
Count pairs of each elements at different indices whose absolute difference equals k
count each pair of indices only once

example: [1,4,3,4,2] k = 2
output: 3 --> (1,3) (5,3) (4,2) //nums hai ye (num1, num2)

```python
arr = [1,4,3,4,2]
k = 2
def count_pairs(arr, k):
    cnt = 0
    n = len(arr)
    for i in range(0,n):
        for j in range(i+1,n):
            abs_diff = abs(arr[i] - arr[j])
            if  abs_diff == k:
                cnt += 1         
            
    return cnt

print(count_pairs(arr, k))
    
```

Find the Number of triplets with a Givern SUm

PRoblem statement: COunt all tripletes at three distinct indices whose sum equals a target value

Example: [1,2,3,4,5] target = 9
output: (1,3,5); (2,3,4) -> 2

```python
arr = [1,2,3,4,5]
target = 9
n = len(arr)

def triplets_sum_equals_target(arr, target):
    cnt = 0
    for i in range(0, n):
        for j in range(i+1,n):
            for k in range(j+1, n):
                if(arr[i] + arr[j] + arr[k] == target):
                    cnt += 1
    return cnt

triplets_sum_equals_target(arr, target)
```

Find the Longest MOuntain in an Array

PRoblem Statement:
A Mountain s a contigous sequence that strictly to a peak and then striclt decreases
Find the length of the longest mountain'a valid mountain must contain at least 3 elements.

Example: [2,1,4,7,3,2,5]
Output: -> [1,4,7,3,2] --> 5 

```python
# Logic -> left se increasing ascending sort then peak mountain idx > left and right and then decreasing
```

```python

def longest_mountain(arr):
    n = len(arr)
    longest = 0

    for i in range(1, n - 1):

        # Check whether i is a peak
        if arr[i - 1] < arr[i] > arr[i + 1]:

            left = i
            right = i

            # Expand towards the left
            while left > 0 and arr[left - 1] < arr[left]:
                left -= 1

            # Expand towards the right
            while right < n - 1 and arr[right] > arr[right + 1]:
                right += 1

            # Calculate mountain length
            length = right - left + 1

            longest = max(longest, length)

    return longest


arr = [2, 1, 4, 7, 3, 2, 5]
print(longest_mountain(arr))  # 5
```
