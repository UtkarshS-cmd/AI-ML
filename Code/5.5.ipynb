Check whether  the array is sorted

Determine wether an array is sorted in ascending order

input: 10 20 25 30 45

outpiut: true

```python
arr = [10, 20, 25, 30, 45]

def sorted_asc(arr):
    for i in range(len(arr) - 1):
        if arr[i] >= arr[i+1]:
            return "not sorted"
        
    return "sorted"


sorted_asc(arr)
        
```
