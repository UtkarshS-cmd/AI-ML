```python
# Find the longest word in a sentence
# Problem statement: Given a sentence containing words separated by spaces,
# print the longest word. If multiple words have the same maximum length, print the first one.
# Example: "Python makes coding enjoyable"
# Output: 'enjoyable'
```

```python
sentence = "Python makes coding enjoyable"
words = sentence.split()
print(words)
longest = words[0]
for word in words:
   if len(word) > len(longest):
       longest = word
print(longest)
```

```python
# Count uppercase and lowercase letters
# Problem statement: Given a string containing English letters,
# count the uppercase letters and lowercase letters separately.
# Ignore digits, spaces, and punctuation.
# Example: "Hello WORLD 123"
# Output:
# upper-case - 6
# lower-case - 4
s = "Hello WORLD 123"
upper = 0
lower = 0
for ch in s:
    if 'A' <= ch <= 'Z':
        upper += 1
    elif 'a' <= ch <= 'z':
        lower += 1
print("Uppercase:", upper)
print("Lowercase:", lower)
Uppercase: 6
Lowercase: 4
```

Toggle the case of every letter
Problem statement:
Convert uppercase letters to lowercase and lowercase letters to uppercase.
Leave digits, spaces, and punctuation unchanged.
Example: "PyThOn 123"
Output: "pYtHoN 123"


```python
s = "PyThOn 123"
result = ""
for ch in s:
    if 'a' <= ch <= 'z':
        result += chr(ord(ch) - 32)
    elif 'A' <= ch <= 'Z':
        result += chr(ord(ch) + 32)
print(result)
```
