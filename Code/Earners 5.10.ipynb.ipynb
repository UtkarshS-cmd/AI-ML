```python
# Problem statement: Given a string containing English letters, 
# count the uppercase letters and lowercase letters separately. 
# Ignore digits, spaces, and punctuation.
```

```python
# Example: "Hello WORLD 123"
# Output: 
# upper-case - 6 
# lower-case - 4
```

```python
s = "Hello WORLD 123"
upper = 0
lower = 0
```

```python
for ch in s:
    if 'A' <= ch <= 'Z':
        upper += 1
    elif 'a' <= ch <= 'z':
        lower += 1
```

```python
print("Uppercase:", upper)
print("Lowercase:", lower)
```

Lowercase: 4

Problem statement: 
Convert uppercase letters to lowercase and lowercase letters to uppercase. 
Leave digits, spaces, and punctuation unchanged.

Example: "PyThOn 123"

```python
s = "PyThOn 123"
result = ""
```

```python
for ch in s:
    if 'a' <= ch <= 'z':
        result += chr(ord(ch) - 32)
    elif 'A' <= ch <= 'Z':
        result += chr(ord(ch) + 32)
```
