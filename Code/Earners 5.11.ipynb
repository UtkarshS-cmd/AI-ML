```python
# Toggle the case of every letter
# Problem statement: Convert uppercase letters to lowercase and lowercase letters to uppercase.
# Leave digits, spaces, and punctuation unchanged.
# Example: "PyThOn 123"
# Output: pYtHoN 123
```

```python
s = "PyThOn 123"
result = ""
for ch in s:
   if 'a' <= ch <= 'z':
     result += chr(ord(ch) - 32)
   elif 'A' <= ch <= 'Z':
     result += chr(ord(ch) + 32)
   else:
       result += ch
print(result)
pYtHoN 123
```

```python
# Replace every vowel with a star
# Problem statement:
# Replace each vowel (a, e, i, o, u, in either case) with *.
# Keep all other characters unchanged.
# Example: "Education"
# Output: "*d*c*t**n"
```

```python
s = "Education"
vowels = "aAeEiIoOuU"
result = ""
for ch in s:
   if ch in vowels:
     result += '*'
   else:
       result += ch
print(result)
*d*c*t**n
```

```python
# # Find the common prefix of two strings
# Problem statement:
# Given two strings, find their longest common prefix:
# the longest sequence of characters they share from the beginning.
# Example: "computer" and "compact"
# Output: "comp"
```

```python
s1 = "computer"
s2 = "compact"
i = 0
n = len(s1)
m = len(s2)
while i < n and i < m:
    if s1[i] != s2[i]:
        break
    i += 1
print(s1[:i])
```

```python
# Check whether one string is a prefix of another
# Problem statement: Given two strings,
# determine whether the first string appears
# exactly at the beginning of the second string.
# Example: First string = "data", second string = "database"
# Output: yes
```

```python
prefix = "data"
text = "database"
if len(prefix) <= len(text) and text[:len(prefix)] == prefix:
    print("Yes")
else:
    print("No")
Yes
```

```python
# Create an array containing only positive elements
# Problem: Create a new array containing only elements greater than zero,
# preserving their original order.
# Example: [-2, 5, 0, 8, -1, 3]
# Output: [5,8,3]
```

```python
arr = [-2, 5, 0, 8, -1, 3]
result = []
for x in arr:
    if x > 0:
        result.append(x)
print(result)
[5, 8, 3]
```

```python
# Check whether all elements are different
# Problem: Determine whether every element in an array is unique.
# Use nested loops without using a set.
# Example: [4, 7, 2, 9]
# Output: True
```

```python
arr = [4, 7, 2, 9]
unique = True
n = len(arr)
for i in range(n):
    for j in range(i + 1, n):
        if arr[i] == arr[j]:
            unique = False
            break
    if not unique:
        break
print("Yes" if unique else "No")
Yes
```
