```python
# Problem statement: 
# Given two strings, find their longest common prefix: 
# the longest sequence of characters they share from the beginning.
```

```python
# Example: "computer" and "compact"
# Output: "comp"
```

```python
s1 = "computer"
s2 = "compact"
i = 0
n = len(s1)
m = len(s2)
```

```python
while i < n and i < m:
    if s1[i] != s2[i]:
        break
        
    i += 1
```

```python
# Problem: Determine whether every element in an array is unique. 
# Use nested loops without using a set.
# Example: [4, 7, 2, 9]
# Output: True
```

```python
arr = [4, 7, 2, 9]
unique = True
n = len(arr)
```

```python
for i in range(n):
    for j in range(i + 1, n):
        if arr[i] == arr[j]:
            unique = False
            break
```

```python
if not unique:
        break
```

```python
print("Yes" if unique else "No")
```

Yes
