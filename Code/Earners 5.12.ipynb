```python
# Find the index of the first element divisible by K
# Problem:
# Find the zero-based index of the first element divisible by K.
# If no such element exists, print -1. Assume K is nonzero.
# Example: [7, 11, 18, 24, 30], K = 6
# Output: 2
```

```python
arr = [7, 11, 18, 24, 30]
k = 6
answer = -1
for i in range(len(arr)):
   if arr[i] % k == 0:
       answer = i
       break
print(answer)
2
```

```python
# Count elements equal to the average
# Problem: Given an array of integers,
# count how many elements equal the array's arithmetic mean.
# If the mean is not an integer, the answer is zero.
# Example: [2, 4, 6, 8]
# Output: 0
# Example: [2, 4, 6]
# Output: 1
```

```python
arr = [2, 4, 6]
total = sum(arr)
n = len(arr)
count = 0
if total % n == 0:
    average = total // n
    for x in arr:
        if x == average:
            count += 1
print(count)
1
```
