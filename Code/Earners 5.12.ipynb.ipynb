```python
# Count elements equal to the average
```

```python
# Problem: Given an array of integers, 
# count how many elements equal the array's arithmetic mean. 
# If the mean is not an integer, the answer is zero.
```

```python
# Example: [2, 4, 6, 8]
# Output: 0
```

```python
# Example: [2, 4, 6]
# Output: 1
```

```python
arr = [2, 4, 6]
total = sum(arr)
n = len(arr)
```

```python
count = 0
```

```python
if total % n == 0:
    average = total // n
```

```python
for x in arr:
        if x == average:
            count += 1
```

```python
print(count)
```

1
