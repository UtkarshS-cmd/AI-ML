```python
# Problem: An array contains N distinct numbers from the range 0 to N,
# with one number missing. Find the missing number.
# Example: [3, 0, 1]
# Output: 2
```

```python
arr = [3, 0, 1]
n = len(arr)
expected = n * (n + 1) // 2
actual = sum(arr)
print(expected - actual)
2
```

```python
# Problem: Given a sorted array of distinct integers and a target,
# return its index if found.
# Otherwise, return the index where it should be inserted to maintain sorted order.
# Example: arr = [1, 3, 5, 6], target = 2 → 1
```

```python
arr=[1,3,5,6]
target=2
def find_index_or_insert(arr,target):
  for i in range(len(arr)):
    if arr[i]==target:
      return i
    elif arr[i]>target:
        arr.insert(i,target)
        return i
find_index_or_insert(arr,target)
# TC: O(n)
1
```

```python
#binary search
arr = [1, 3, 5, 6]
target = 2
left = 0
right = len(arr) - 1
while left <= right:
    mid = (left + right) // 2
    if arr[mid] == target:
        print(mid)
        break
    elif arr[mid] < target:
        left = mid + 1
    else:
        right = mid - 1
else:
    print(left)
# TC: O(logn)
1
```

```python
# Find the maximum number of consecutive ones
# Problem: Given a binary array containing only 0 and 1,
# find the maximum number of consecutive 1s.
# Example: [1, 1, 0, 1, 1, 1]
# Output: 3
```

```python
arr = [1, 1, 0, 1, 1, 1]
current = 0
mx = 0
for x in arr:
    if x == 1:
        current += 1
        mx = max(mx, current)
    else:
        current = 0
print(mx)
3
```
