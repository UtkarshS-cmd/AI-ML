```python
# Problem: Given a sorted array of distinct integers and a target, 
# return its index if found. 
# Otherwise, return the index where it should be inserted to maintain sorted order.
```

```python
# Example: arr = [1, 3, 5, 6], target = 2 → 1
```

```python
arr=[1,3,5,6]
target=2
```

```python
def find_index_or_insert(arr,target):
  for i in range(len(arr)):
    if arr[i]==target:
      return i
    elif arr[i]>target:
        arr.insert(i,target)
        return i
  
find_index_or_insert(arr,target)
```

```python
# TC: O(n)
```

```python
#binary search 
arr = [1, 3, 5, 6]
target = 2
```

```python
left = 0
right = len(arr) - 1
```

```python
while left <= right:
    mid = (left + right) // 2
```

```python
if arr[mid] == target:
        print(mid)
        break
    elif arr[mid] < target:
        left = mid + 1
    else:
        right = mid - 1
else:
    print(left)
```

```python
# TC: O(logn)
```

```python
# Problem: Remove all occurrences of a given value from an array in-place. 
# Return the number of remaining elements; 
# the first part of the array should contain the retained elements.
```

```python
# Example: arr = [3, 2, 2, 3], value = 3
# Output: [2,2] len = 2
```

```python
arr = [3, 2, 2, 3]
value = 3
k = 0
```

```python
for x in arr:
    if x != value:
        arr[k] = x
        k += 1
```

```python
print("Length:", k)
print("Remaining:", arr[:k])
```

Length: 2
Remaining: [2, 2]

```python
# Problem: Given a binary array containing only 0 and 1, 
# find the maximum number of consecutive 1s.
```

```python
# Example: [1, 1, 0, 1, 1, 1]
# Output: 3
```

```python
arr = [1, 1, 0, 1, 1, 1]
current = 0
mx = 0
```

```python
for x in arr:
    if x == 1:
        current += 1
        mx = max(mx, current)
    else:
        current = 0
```

```python
print(mx)
```
