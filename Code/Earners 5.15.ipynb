```python
# Problem: String t is created by adding one character to string s
# and rearranging the characters. Find the added character.
# Example: s = "abcd", t = "abcde"
# Output: 'e'
```

```python
s = "abcd"
t = "abcde"
answer = 0
for ch in s:
    answer ^= ord(ch)
for ch in t:
    answer ^= ord(ch)
print(chr(answer))
e
```

```python
# Problem: Determine whether all characters of string s appear in string t in the same order,
# though not necessarily consecutively.
# Example: s = "ace", t = "abcde"
# Output: Yes
```

```python
s = "ace"
t = "abcde"
i = 0
j = 0
while i < len(s) and j < len(t):
    if s[i] == t[j]:
        i += 1
    j += 1
print(i == len(s))
```

```python
# Problem: Given a non-empty string, find the character that occurs most frequently.
# If multiple characters have the same highest frequency,
# return the one that appears first in the string.
# Example: "success"
# Output: s
```

```python
s = "success"
freq = {}
for ch in s:
    freq[ch] = freq.get(ch, 0) + 1
answer = s[0]
for ch in s:
    if freq[ch] > freq[answer]:
        answer = ch
print(answer)
s
```

Problem: Given two lowercase strings, calculate how many character additions or removals are required
to make their character frequencies equal.
Each insertion or deletion changes one character.
Example: s = "abc", t = "aab" → 2
abc
aab --> remove:a , add c --> acb


```python
s = "abc"
t = "aab"
freq = [0] * 26
for ch in s:
   freq[ord(ch) - ord('a')] += 1
for ch in t:
   freq[ord(ch) - ord('a')] -= 1
operations = 0
for count in freq:
   operations += abs(count)
print(operations)
2
```
