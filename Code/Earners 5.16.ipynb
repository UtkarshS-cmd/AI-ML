```python
# Given an array of integers, reverse its elements using a stack.
# You must first push all array elements into the stack and
# then retrieve them one by one to construct the reversed array.
# Input:
# [10, 20, 30, 40, 50]
# Output:
# [50, 40, 30, 20, 10]
```

```python
def reverse_array(arr):
    stack = []
    for x in arr:
        stack.append(x)
    result = []
    while stack:
        result.append(stack.pop())
    return result
arr = [10, 20, 30, 40, 50]
print(reverse_array(arr))
[50, 40, 30, 20, 10]
```

```python
# Given two integer arrays, return two lists:
# - Elements present in the first array but not in the second.
# - Elements present in the second array but not in the first.
# Each element should appear only once in its result list.
# Example:
# Input:
# nums1 = [1, 2, 3, 3]
# nums2 = [2, 4, 5]
# Output: [[1,3], [4,5]]
```

```python
def find_difference(nums1, nums2):
    set1 = set(nums1)
    set2 = set(nums2)
    first = list(set1 - set2)
    second = list(set2 - set1)
    return [first, second]
print(find_difference([1, 2, 3, 3], [2, 4, 5]))
```

```python
# Given a positive integer N, find an integer X such that:
# - Sum of all integers from 1 to X equals the sum of all integers from X to N.
# Return X. If no such integer exists, return -1.
# Input:
# N = 8
# Output: 6
# 1-6: 21
# 6-8: 21
```

```python
# sum of 1st n natural numbers: n(n+1)/2
# 1-n --> n(n+1)/2
# x : 1 to n
# 1 2 3 4 5 6
# sum(6)
# sum(3)
# ans = sum(6)-sum(3)
```

```python
def pivot_integer(n):
    total = n * (n + 1) // 2
    for x in range(1, n + 1):
        left = x * (x + 1) // 2
        right = total - left + x
        if left == right:
            return x
    return -1
print(pivot_integer(8))
6
```

Given a list of sentences,
find the maximum number of words present in any single sentence.
Words are separated by spaces.
[
 "I love programming",
 "Data structures are important",
 "Practice daily"
]


```python
def maximum_words(sentences):
    mx = 0
    for sentence in sentences:
        words = sentence.split()
        print(len(words))
        mx = max(mx, len(words))
    return mx
sentences = [
    "I love programming",
    "Data structures are important",
    "Practice daily"
]
print(maximum_words(sentences))
3
4
2
4
```

```python
# odd --> odd : 0
# even --> even : 0
position = [1, 2, 3]
even = 0
odd = 0
for p in position:
    if p % 2 == 0:
        even += 1
    else:
        odd += 1
ans = min(even, odd )
print(ans)
1
```

```python
# Given a 2D matrix containing integers, calculate the sum of all its elements.
# Input:
# [
#  [1, 2, 3],
#  [4, 5, 6]
# ]
# Output: 21
```

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]
total = 0
for row in matrix:
    for value in row:
        total += value
print(total)
21
```

Given an M × N matrix, return its transpose.


```python
matrix = [
    [1, 2, 3],
    [4, 5, 6]
]
rows = len(matrix)
cols = len(matrix[0])
for j in range(cols):
    for i in range(rows):
        print(matrix[i][j], end = " ")
    print()
1 4
2 5
3 6
```

```python
def transpose(matrix):
   rows = len(matrix)
   cols = len(matrix[0])
   result = []
   for j in range(cols):
       new_row = []
       for i in range(rows):
           new_row.append(matrix[i][j])
       result.append(new_row)
   return result
matrix = [
   [1, 2, 3],
   [4, 5, 6]
]
print(transpose(matrix))
[[1, 4], [2, 5], [3, 6]]
```
