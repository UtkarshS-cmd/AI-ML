```python
# Given an array of integers, reverse its elements using a stack.
```

```python
# You must first push all array elements into the stack and 
# then retrieve them one by one to construct the reversed array.
```

```python
# Input:
# [10, 20, 30, 40, 50]
```

```python
# Output:
# [50, 40, 30, 20, 10]
```

```python
def reverse_array(arr):
    stack = []
```

```python
for x in arr:
        stack.append(x)
```

```python
result = []
```

```python
while stack:
        result.append(stack.pop())
```

```python
return result
```

```python
arr = [10, 20, 30, 40, 50]
print(reverse_array(arr))
```

[50, 40, 30, 20, 10]

```python
# Given a positive integer N, find an integer X such that:
```

```python
# - Sum of all integers from 1 to X equals the sum of all integers from X to N.
```

```python
# Return X. If no such integer exists, return -1.
```

```python
# Input:
# N = 8
```

```python
# Output: 6
```

```python
# 1-6: 21 
# 6-8: 21
```

```python
# sum of 1st n natural numbers: n(n+1)/2
```

```python
# 1-n --> n(n+1)/2
```

```python
# x : 1 to n
```

```python
# 1 2 3 4 5 6
```

```python
# sum(6)
# sum(3)
# ans = sum(6)-sum(3)
```

```python
def pivot_integer(n):
    total = n * (n + 1) // 2
```

```python
for x in range(1, n + 1):
        left = x * (x + 1) // 2
        right = total - left + x
```

```python
if left == right:
            return x
```

```python
return -1
```

```python
print(pivot_integer(8))
```

```python
# You can move a chip:
# - By 2 positions at a cost of 0.
# - By 1 position at a cost of 1.
# Find the minimum total cost required to move all chips to the same position.
```

```python
# Input:
# position = [1, 2, 3]
# Output: 0+1 = 1
```

```python
# odd --> odd : 0 
# even --> even : 0
```

```python
position = [1, 2, 3]
```

```python
even = 0
odd = 0
```

```python
for p in position:
    if p % 2 == 0:
        even += 1
    else:
        odd += 1
```

```python
ans = min(even, odd )
```

```python
print(ans)
```
