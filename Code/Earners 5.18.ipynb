```python
# You are selling lemonade for ₹5 per glass.
# Customers pay using ₹5, ₹10, or ₹20 notes. Each customer buys exactly one glass.
# Initially, you have no change.
# Determine whether you can provide the correct change to every customer in the given order.
# Input:
# bills = [5, 5, 5, 10, 20]
# Output:
# True
5 -> add 5
10 --> -5, add 10
20 --> 10+5, 5*3
```

```python
def lemonade_change(bills):
    five = 0
    ten = 0
    for bill in bills:
        if bill == 5:
            five += 1
        elif bill == 10:
            if five == 0:
                return False
            five -= 1
            ten += 1
        else:
            if ten > 0 and five > 0:
                ten -= 1
                five -= 1
            elif five >= 3:
                five -= 3
            else:
                return False
    return True
print(lemonade_change([5, 5, 5, 10, 20]))
True
```

```python
# Problem statement: Given a binary array containing only 0s and 1s,
# delete exactly one element. Return the maximum number of consecutive 1s remaining.
# Input: nums = [1, 1, 0, 1]
# Output: 3
```

```python
def longest_ones(nums):
    left = 0
    zeros = 0
    ans = 0
    for right in range(len(nums)):
        if nums[right] == 0:
            zeros += 1
        while zeros > 1:
            if nums[left] == 0:
                zeros -= 1
            left += 1
        ans = max(ans, right - left)
    return ans
print(longest_ones([1, 1, 0, 1]))
3
```

```python
# A sorted array has been rotated at some pivot.
# Find the index of its smallest element.
# Assume all elements are distinct.
# Input: nums = [4, 5, 6, 7, 0, 1, 2]
# Output: 4
```

```python
def find_min_index(nums):
    left = 0
    right = len(nums) - 1
    while left < right:
        mid = (left + right) // 2
        if nums[mid] > nums[right]:
            left = mid + 1
        else:
            right = mid
```

```python
# Problem statement: Given a sorted array and a target value,
# return the starting and ending positions of that target.
# If the target does not exist, return [-1, -1].
# Input: nums = [5, 7, 7, 8, 8, 10], target = 8
# Output: [3,4]
```

```python
def search_range(nums, target):
    first = -1
    last = -1
    for i in range(len(nums)):
        if nums[i] == target:
            if first == -1:
                first = i
            last = i
    return [first, last]
print(search_range([5, 7, 7, 8, 8,8,8, 10], 8))
[3, 6]
```
