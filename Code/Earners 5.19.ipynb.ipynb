```python
# Problem statement: Given an integer array and an integer k, 
# find the contiguous subarray of length k having the maximum average. 
# Return the maximum average.
```

```python
# Input: nums = [1, 12, -5, -6, 50, 3], k = 4
# output:
# Subarray [12, -5, -6, 50]
# sum = 51
# avg = 51/4 = 12.75
```

```python
def find_max_average(nums, k):
    current_sum = sum(nums[:k])
    max_sum = current_sum
```

```python
for i in range(k, len(nums)):
        current_sum += nums[i] - nums[i - k]
        max_sum = max(max_sum, current_sum)
```

```python
return max_sum / k
```

```python
print(find_max_average([1, 12, -5, -6, 50, 3], 4))
```

12.75

```python
# Problem statement: 
# Given an array of positive integers and a target average T, 
# find the length of the smallest contiguous subarray whose average is at least T. 
# Return 0 if no such subarray exists.
```

```python
# Input: nums = [2, 3, 1, 2, 4, 3], T = 3
# ans = 1
```

```python
def min_subarray_length(nums, T):
    left = 0
    total = 0
    ans = float('inf')
```

```python
for right in range(len(nums)):
        total += nums[right]
```

```python
while left <= right and total >= T:
            ans = min(ans, right - left + 1)
            total -= nums[left]
            left += 1
```

```python
return 0 if ans == float('inf') else ans
```

```python
print(min_subarray_length([2, 3, 1, 2, 4, 3], 4))
```

1

```python
# Problem statement: Given a sentence, 
# reverse every individual word without changing the order of the words.
```

```python
# Input: s = "Hello World Python"
# Output: "olleH dlroW nohtyP"
```

```python
def reverse_each_word(s):
    words = s.split()
    result = []
```

```python
for word in words:
        result.append(word[::-1])
```

```python
return " ".join(result)
```

```python
print(reverse_each_word("Hello World Python"))
```

olleH dlroW nohtyP
