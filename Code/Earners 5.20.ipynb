```python
# Problem: Merge two sorted arrays into the first array.
# The first array has enough empty space at the end.
# Input:
# nums1 = [1, 2, 3, 0, 0, 0]
# nums2 = [2, 5, 6]
# Output: [1, 2, 2, 3, 5, 6]
```

```python
def merge(nums1, m, nums2, n):
    i = m - 1
    j = n - 1
    k = m + n - 1
    while j >= 0:
        if i >= 0 and nums1[i] > nums2[j]:
            nums1[k] = nums1[i]
            i -= 1
        else:
            nums1[k] = nums2[j]
            j -= 1
        k -= 1
    return nums1
print(merge([1, 2, 3, 0, 0, 0], 3, [2, 5, 6], 3))
[1, 2, 2, 3, 5, 6]
```

```python
arr1 = [1, 2, 3, 0, 0, 0]
arr2 = [2, 5, 6]
def merge_arrays(arr1, arr2):
    j = 0
    for i in range(len(arr1)):
        if arr1[i] == 0 and j < len(arr2):
            arr1[i] = arr2[j]
            j += 1
    for i in range(len(arr1)):
        for j in range(i + 1, len(arr1)):
            if arr1[i] > arr1[j]:
                arr1[i], arr1[j] = arr1[j], arr1[i]
    return arr1
print(merge_arrays(arr1, arr2))
[1, 2, 2, 3, 5, 6]
```

```python
# Determine whether an array contains any duplicate element.
# Example: [1, 2, 3, 1]
# Output: True
```

```python
def contains_duplicate(nums):
    seen = set()
    for num in nums:
        if num in seen:
            return True
        seen.add(num)
    return False
print(contains_duplicate([1, 2, 3, 1]))
True
```

```python
# Find the indices of two elements whose sum equals the target.
# Example: [2, 7, 11, 15], target = 9
# Output: [0,1]
# Max 1 loop
```

```python
def two_sum(nums, target):
    seen = {}
    for i in range(len(nums)):
        complement = target - nums[i]
        if complement in seen:
            return [seen[complement], i]
        seen[nums[i]] = i
    return []
print(two_sum([2, 7, 11, 15], 9))
# 9
# 2
# complement = 9-2 = 7
[0, 1]
```

```python
def find_min(nums):
    left = 0
    right = len(nums) - 1
    while left < right:
        mid = (left + right) // 2
        if nums[mid] > nums[right]:
            left = mid + 1
        else:
            right = mid
    return nums[left]
print(find_min([3, 4, 5, 1, 2]))
1
```
