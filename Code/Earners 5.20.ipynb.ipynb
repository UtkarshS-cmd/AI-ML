```python
# Problem: Merge two sorted arrays into the first array. 
# The first array has enough empty space at the end.
```

```python
# Input:
# nums1 = [1, 2, 3, 0, 0, 0]
# nums2 = [2, 5, 6]
```

```python
# Output: [1, 2, 2, 3, 5, 6]
```

```python
def merge(nums1, m, nums2, n):
    i = m - 1
    j = n - 1
    k = m + n - 1
```

```python
while j >= 0:
        if i >= 0 and nums1[i] > nums2[j]:
            nums1[k] = nums1[i]
            i -= 1
        else:
            nums1[k] = nums2[j]
            j -= 1
```

k -= 1

```python
return nums1
```

```python
print(merge([1, 2, 3, 0, 0, 0], 3, [2, 5, 6], 3))
```

```python
arr1 = [1, 2, 3, 0, 0, 0]
arr2 = [2, 5, 6]
```

```python
def merge_arrays(arr1, arr2):
    j = 0
```

```python
for i in range(len(arr1)):
        if arr1[i] == 0 and j < len(arr2):
            arr1[i] = arr2[j]
            j += 1
```

```python
for i in range(len(arr1)):
        for j in range(i + 1, len(arr1)):
            if arr1[i] > arr1[j]:
                arr1[i], arr1[j] = arr1[j], arr1[i]
```

```python
return arr1
```

```python
print(merge_arrays(arr1, arr2))
```

```python
# Example: [1, 2, 3, 4]
# Output: [24, 12, 8, 6]
```

```python
def product_except_self(nums):
    n = len(nums)
    result = [1] * n
```

```python
left_product = 1
```

```python
for i in range(n):
        result[i] = left_product
        left_product *= nums[i]
```

```python
# Determine whether an array contains any duplicate element.
```

```python
# Example: [1, 2, 3, 1]
# Output: True
```

```python
def contains_duplicate(nums):
    seen = set()
```

```python
for num in nums:
        if num in seen:
            return True
```

seen.add(num)

```python
return False
```

```python
print(contains_duplicate([1, 2, 3, 1]))
```

```python
def find_min(nums):
    left = 0
    right = len(nums) - 1
```

```python
while left < right:
        mid = (left + right) // 2
```

```python
if nums[mid] > nums[right]:
            left = mid + 1
        else:
            right = mid
```

```python
return nums[left]
```

```python
print(find_min([3, 4, 5, 1, 2]))
```
