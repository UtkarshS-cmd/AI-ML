Count the Number of Inversions
Problem statement: An inversion is a pair of indices (i, j) where i < j but arr[i] > arr[j].
Count all inversions in the array.
Example: [2, 4, 1, 3, 5]
Output: (2,1), (4,1) (4,3) --> 3


```python
arr=[2,4,1,3,5]
def find_inversion_pair(arr):
 count=0
 for i in range(len(arr)):
   for j in range(i+1,len(arr)):
     if arr[i]>arr[j]:
       count+=1
       print(arr[i], arr[j])
  return count
find_inversion_pair(arr)
```

Problem: Find the greatest common divisor (GCD) of two positive integers
using the Euclidean algorithm.
input: 48, 18
Output:
18 --> 1, 2, 3, 6, 9, 18
48 --> 1, 2, 3, 4, 6, 8, 12, 16, 24, 48
common divisors --> 1, 2, 3, 6
greatest common divisor ==> 6


```python
input = [18, 48]
def find_gcd(arr):
 greatest_divisor = 1
 for i in range(1, min(arr) + 1):
   if arr[0] % i == 0 and arr[1] % i == 0:
       greatest_divisor = i
  return greatest_divisor
find_gcd(input)
```

```python
# Given two integers L and R, print all prime numbers in the inclusive range.
# Example: Input: 10, 25
# Output: 11, 13, 17, 19, 23
L, R = 10, 25
for n in range(max(2, L), R + 1):
    prime = True
    for d in range(2, int(n ** 0.5) + 1):
        if n % d == 0:
            prime = False
            break
    if prime:
        print(n, end=" ")
```

```python
matrix = [
    [1, 4, 7],
    [2, 5, 8],
    [3, 6, 9]
]
target = 6
rows = len(matrix)
cols = len(matrix[0])
i, j = 0, cols - 1
found = False
while i < rows and j >= 0:
    if matrix[i][j] == target:
        found = True
        break
    elif matrix[i][j] > target:
        j -= 1
    else:
        i += 1
print("Found" if found else "Not Found")
```
