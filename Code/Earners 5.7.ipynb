```python
# Repeatedly add the digits of a positive integer until only one digit remains.
# Input: 9875
# Output:
# 9875 --> 9 + 8 + 7 + 5 = 29
# 29 --> 2 + 9 = 11
# 11 --> 1 + 1 = 2
n = 9875
while n >= 10:
   digit_sum = 0
   while n > 0:
       digit_sum += n % 10
       n //= 10
   n = digit_sum
print(n)
2
```

```python
# A number is an Armstrong number
# if the sum of each digit raised to the number of digits equals the original number.
# n = 153
# d = 3
# 1^3 + 5^3 + 3^3 = 1 + 125 + 27 = 153 == 153 --> armstrong
```

```python
n = 153
d = count_digits(n)
sm = 0
temp = n
while n>0:
  digit = n%10
  n //= 10
  sm += digit**d
if sm == temp:
  print("Armstrong")
else:
  print("Not Armstrong")
Armstrong
```

```python
# Find the Nth Fibonacci number
# Problem:
# The Fibonacci sequence starts with 0, 1.
# Each subsequent term is the sum of the previous two.
# Find the term at zero-based index N.
0, 1, 1, 2, 3, 5, 8, 13, 21 ...
# n = 7, output: 13
```
