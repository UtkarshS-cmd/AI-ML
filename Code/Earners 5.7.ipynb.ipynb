```python
# A number is an Armstrong number 
# if the sum of each digit raised to the number of digits equals the original number.
```

```python
# n = 153 
# d = 3
```

```python
# 1^3 + 5^3 + 3^3 = 1 + 125 + 27 = 153 == 153 --> armstrong
```

```python
n = 153 
d = count_digits(n)
sm = 0
temp = n
```

```python
while n>0:
  digit = n%10 
  n //= 10 
  sm += digit**d
```

```python
if sm == temp:
  print("Armstrong")
else:
  print("Not Armstrong")
```

Armstrong

```python
def fact(n):
  if n <= 1:
    return 1
  
  return n*fact(n-1)
```

```python
# Problem: 
# The Fibonacci sequence starts with 0, 1. 
# Each subsequent term is the sum of the previous two. 
# Find the term at zero-based index N.
```

```python
0, 1, 1, 2, 3, 5, 8, 13, 21 ...
# n = 7, output: 13
```

```python
prev = 0 
curr = 1 
temp = 0
n = 7
for i in range(2, n+1):
  temp = curr + prev 
  prev = curr
  curr = temp
```

```python
print(temp)
```

13
