```python
s1 = "listen"
s2 = "silent"
```

```python
if len(s1) != len(s2):
    print("Not Anagrams")
else:
    freq = [0] * 26
```

```python
for ch in s1:
        freq[ord(ch) - ord('a')] += 1
```

```python
for ch in s2:
        freq[ord(ch) - ord('a')] -= 1
```

```python
if all(x == 0 for x in freq):
        print("Anagrams")
    else:
        print("Not Anagrams")
```

```python
# Calculate the average of elements greater than a given value
```

```python
# Problem statement: 
# Given an array and a number K, calculate the average of all elements strictly greater than K. 
# If no element qualifies, print No elements.
```

```python
# Example: Array = [10, 25, 30, 15, 40], K = 20
# Output: 25, 30, 40 --> 95/3 = 31.67
```

```python
arr = [10, 25, 30, 15, 40]
k = 20
```

```python
total = 0
count = 0
```

```python
for x in arr:
    if x > k:
        total += x
        count += 1
```

```python
if count > 0:
    print(round(total / count, 2))
else:
    print("No elements")
```

```python
# Find the difference between the sum of even and odd elements
```

```python
# Problem statement: Calculate the sum of even-valued elements and the sum of odd-valued elements. 
# Print the absolute difference between the two sums.
```

```python
# Example: [2, 7, 4, 9, 6]
# Output: 
# even-sum = 2 + 4 + 6 = 12 
# odd-sum = 7+9 = 16
```

```python
# abs(diff) = 16-12 = 4
```

```python
a = [2, 7, 4, 9, 6]
```

```python
even_sum = 0
odd_sum = 0
```

```python
for elem in a:
  if elem%2 == 0:
    even_sum += elem 
  else:
    odd_sum += elem
```

```python
print(even_sum, odd_sum)
diff = abs(even_sum-odd_sum)
print(diff)
```
