```python
# Find the difference between the sum of even and odd elements
# Problem statement: Calculate the sum of even-valued elements and the sum of odd-valued elements.
# Print the absolute difference between the two sums.
# Example: [2, 7, 4, 9, 6]
# Output: 4
# sum_even = 2 + 4 + 6 = 12
# sum_odd = 7+9 = 16
# diff = 16-12 = 4
```

```python
arr = [2,7,4,9,6]
even_sum = 0
odd_sum = 0
for num in arr:
 if num % 2 == 0:
   even_sum += num
 else:
   odd_sum += num
print(even_sum , odd_sum)
diff = abs(even_sum - odd_sum)
print(diff)
```

```python
# Find the element closest to zero
# Problem statement: Given an array containing positive and negative integers,
# find the element whose absolute value is smallest.
# If two elements are equally close to zero, choose the positive one.
# Example: [-8, 3, -2, 5, 2]
# Output: 2
```

```python
def solve(arr):
    closest = abs(arr[0])
    elem = arr[0]
    n = len(arr)
    for i in range(1, n):
        if abs(arr[i]) < closest:
            closest = abs(arr[i])
            elem = arr[i]
        elif abs(arr[i]) == closest and arr[i] > elem:
            elem = arr[i]
    return elem
result = solve([-8, 3, -2, 5, 2])
print(result)
```

```python
# Count elements that are divisible by both 3 and 5
# Problem statement: Count how many elements in an array are divisible by both 3 and 5.
# Example: [15, 10, 30, 9, 45, 7]
# Output: 3 --> 15, 30, 45
```

```python
arr = [15, 10, 30, 9, 45, 7]
count = 0
for x in arr:
    if x % 3 == 0 and x % 5 == 0:
        count += 1
        print(x)
```

```python
# Find the longest consecutive run of positive numbers
# Problem statement:
# Find the length of the longest contiguous sequence containing only positive numbers.
# Zero and negative values break the sequence.
# Example: [2, 3, -1, 4, 5, 6, 0, 8]
# Output: 3
```

```python
arr = [2, 3, -1, 4, 5, 6, 0, 8]
current = 0
longest = 0
for x in arr:
   if x > 0:
       current += 1
       longest = max(longest, current)
   else:
       current = 0
print(longest)
```

```python
# Interleave two arrays
# Problem statement: Given two arrays of equal length,
# create a new array by alternating elements from each array, starting with the first array.
# Example: A = [1, 2, 3], B = [7, 8, 9]
# Output: [1, 7, 2, 8, 3, 9]
```

```python
a = [1, 2, 3]
b = [7, 8, 9]
result = []
for i in range(len(a)):
    result.append(a[i])
    result.append(b[i])
print(result)
[1, 7, 2, 8, 3, 9]
```
